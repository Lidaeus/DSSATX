# 2026-10-07预审的数值复核
下列只复核已保存的上下文与均值/影响计算，人工科学判断见逐项说明。不执行源数据或核对表写入。

```python
import json, statistics, math
from pathlib import Path
OUT=Path(r'D:\三季节数据汇总\数据质控_20261007\预审_20261007')
d=json.loads((OUT/'逐项预审建议.json').read_text(encoding='utf8'))
for c in d['cases']:
    m=statistics.mean(c['peer_values'])
    assert math.isclose(m,c['peer_mean'],rel_tol=1e-12)
    original=(sum(c['peer_values'])+c['current_value'])/3
    after=(sum(c['peer_values'])+m)/3
    assert math.isclose((after-original)/original,c['replacement_group_impact'],rel_tol=1e-12,abs_tol=1e-12)
print(d['counts_unique'])
```
